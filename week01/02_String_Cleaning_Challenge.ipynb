# String Cleaning Challenge

```python
message = " Welcome To Python Programming Class "
```

#### 1. Remove extra spaces

```python
print(message.strip())
```

#### 2. Convert everything to lowercase

```python
print(message.lower())
```

#### 3. Convert everything to uppercase

```python
print(message.upper())
```

#### 4. Convert to title case

```python
print(message.title())
```

#### 5. Replace "Python" with "Advanced Python"

```python
print(message.replace("Python", "Advanced Python"))
```

#### 6. Check whether the string starts with "Welcome"

```python
print(message.strip().startswith("Welcome"))
```

```python
print(message.strip().startswith("welcome"))
```

#### 7. Check whether the string ends with "Class"

```python
print(message.strip().endswith("Class"))
```

#### 8. Count occurrences of "o"

```python
print(message.count("o"))
```

#### 9. Find the position of "Programming"

```python
print(message.find("Programming"))
```

#### 10. Split the sentence into words

```python
print(message.split())
```
