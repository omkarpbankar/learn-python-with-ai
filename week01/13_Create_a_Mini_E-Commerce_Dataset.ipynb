# Python — List of Dictionaries: Product Dataset

This notebook demonstrates how a **list of dictionaries** can be used to represent structured product data.

## Concepts covered
- List
- Dictionary
- Indexing
- Keys
- Values
- Updating dictionary values
- Adding new dictionary keys
- Adding a new dictionary to a list
- Working without loops


````python
# Create a list containing three product dictionaries

products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell"
    },
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung"
    },
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple"
    }
]

print("Initial product dataset:")
print(products)
````

````python
# 1. Print all products

print("All products:")
print(products)
````

````python
# 2. Print the first product

print("First product:")
print(products[0])
````

````python
# 3. Print the second product's price

print("Second product's price:")
print(products[1]["price"])
````

````python
# 4. Print the third product's brand

print("Third product's brand:")
print(products[2]["brand"])
````

````python
# 5. Change the first product's price

products[0]["price"] = 75000

print("Updated first product:")
print(products[0])
````

````python
# 6. Add "rating" to the second product

products[1]["rating"] = 4.5

print("Updated second product:")
print(products[1])
````

````python
# 7. Add another product manually

products.append({
    "name": "Smartwatch",
    "price": 15000,
    "brand": "Apple"
})

print("After adding another product:")
print(products)
````

````python
# 8. Print the final dataset

print("Final product dataset:")
print(products)
````

# Explanation: List + Dictionary

## 1. List

A list is an ordered collection of values.

In this example:

```python
products = [
    {...},
    {...},
    {...}
]
```

The list stores multiple product dictionaries.

We use **square brackets `[]`** to create a list.

---

## 2. Dictionary

Each product is represented by a dictionary:

```python
{
    "name": "Laptop",
    "price": 70000,
    "brand": "Dell"
}
```

A dictionary stores data as **key-value pairs**.

We use **curly braces `{}`** to create a dictionary.

---

## 3. Indexing

Lists use zero-based indexing.

```python
products[0]  # First product
products[1]  # Second product
products[2]  # Third product
```

The first item has index `0`, not `1`.

---

## 4. Keys

Dictionary keys identify pieces of information.

For example:

```python
"name"
"price"
"brand"
```

To access a value using its key:

```python
products[0]["name"]
```

This first selects the first product and then selects its `"name"` value.

---

## 5. Values

Values are the actual data stored against the keys.

For example:

```python
products[0]["name"]   # "Laptop"
products[0]["price"]  # 70000
products[0]["brand"]  # "Dell"
```

---

## Combining Indexing and Keys

This is the most important pattern in this exercise:

```python
products[index]["key"]
```

For example:

```python
products[1]["price"]
```

Python performs two steps:

1. `products[1]` → selects the second product dictionary.
2. `["price"]` → retrieves the price from that dictionary.

Therefore, the result is:

```text
40000
```

---

## Updating Data

We can change a value using the same pattern:

```python
products[0]["price"] = 75000
```

This changes the first product's price.

---

## Adding a New Key

We can add a new key-value pair:

```python
products[1]["rating"] = 4.5
```

Now the second product contains:

```python
{
    "name": "Phone",
    "price": 40000,
    "brand": "Samsung",
    "rating": 4.5
}
```

---

## Adding Another Product

Because `products` is a list, we can use `append()`:

```python
products.append({
    "name": "Smartwatch",
    "price": 15000,
    "brand": "Apple"
})
```

This adds another dictionary to the list.


````python
# Bonus practice: Access keys and values from individual products

print("First product keys:")
print(products[0].keys())

print("\nFirst product values:")
print(products[0].values())

print("\nSecond product keys:")
print(products[1].keys())

print("\nSecond product values:")
print(products[1].values())
````

## Quick Summary

| Concept | Example | Meaning |
|---|---|---|
| List | `products` | Stores multiple products |
| Dictionary | `products[0]` | Represents one product |
| Indexing | `products[1]` | Selects the second product |
| Key | `"price"` | Identifies a dictionary field |
| Value | `40000` | Data stored under a key |
| Combined access | `products[1]["price"]` | Gets the price of the second product |
