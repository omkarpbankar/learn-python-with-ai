# Python Nested Dictionaries — Employee Record

This notebook demonstrates how to create, access, and modify a **nested dictionary** without using loops.

### Concepts covered
- Dictionary creation
- Nested dictionaries
- Dictionary indexing
- Updating values
- Adding new key-value pairs
- Accessing nested data


````python
# Create the employee nested dictionary

employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

print(employee)
````

````python
# 1. Print employee name

print("Employee name:")
print(employee["name"])
````

````python
# 2. Print department

print("Department:")
print(employee["department"])
````

````python
# 3. Print complete skills dictionary

print("Skills:")
print(employee["skills"])
````

````python
# 4. Print programming language

print("Programming language:")
print(employee["skills"]["language"])
````

````python
# 5. Print database

print("Database:")
print(employee["skills"]["database"])
````

````python
# 6. Print cloud technology

print("Cloud technology:")
print(employee["skills"]["cloud"])
````

````python
# 7. Change Python to Python + JavaScript

employee["skills"]["language"] = "Python + JavaScript"

print("Updated programming language:")
print(employee["skills"]["language"])
````

````python
# 8. Change salary

employee["salary"] = 95000

print("Updated salary:")
print(employee["salary"])
````

````python
# 9. Add experience

employee["experience"] = 3

print("Employee experience:")
print(employee["experience"])
````

````python
# 10. Add another skill under the skills dictionary

employee["skills"]["framework"] = "Django"

print("Updated skills:")
print(employee["skills"])
````

````python
# Print the final employee record

print("Final employee record:")
print(employee)
````

## How to Access Nested Dictionaries

A dictionary can contain another dictionary as a value.

In this example:

```python
employee = {
    "name": "Amit",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    }
}
```

The `skills` key contains another dictionary.

### Accessing the outer dictionary

To access the employee's name:

```python
employee["name"]
```

To access the complete skills dictionary:

```python
employee["skills"]
```

### Accessing the inner dictionary

To access the programming language:

```python
employee["skills"]["language"]
```

Python evaluates this from left to right:

1. `employee["skills"]` → gets the nested skills dictionary
2. `["language"]` → gets the value stored under the `language` key

Similarly:

```python
employee["skills"]["database"]
```

returns:

```text
PostgreSQL
```

And:

```python
employee["skills"]["cloud"]
```

returns:

```text
AWS
```

### Updating nested values

The same indexing pattern can be used to update values:

```python
employee["skills"]["language"] = "Python + JavaScript"
```

### Adding a new nested key

```python
employee["skills"]["framework"] = "Django"
```

This adds a new key-value pair inside the `skills` dictionary.

### General pattern

For a dictionary nested inside another dictionary:

```python
outer_dictionary["outer_key"]["inner_key"]
```

This is called **nested dictionary access**.
