# Python List Operations — Shopping Cart

This exercise demonstrates common Python list methods using a shopping cart.

### Topics covered
- `append()`
- `extend()`
- `insert()`
- `remove()`
- `pop()`
- `copy()`
- `sort()`
- `reverse()`
- Indexing
- `count()`


````python
# Create the shopping cart

cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]

print("Original cart:")
print(cart)
````

````python
# 1. Display all products

print("All products:")
print(cart)
````

````python
# 2. Access first and last products

print("First product:", cart[0])
print("Last product:", cart[-1])
````

````python
# 3. Add "Webcam"

cart.append("Webcam")

print("After adding Webcam:")
print(cart)
````

````python
# 4. Insert "USB Hub" at index 2

cart.insert(2, "USB Hub")

print("After inserting USB Hub at index 2:")
print(cart)
````

````python
# 5. Remove "Mouse"

cart.remove("Mouse")

print("After removing Mouse:")
print(cart)
````

````python
# 6. Remove the last item using pop()

removed_item = cart.pop()

print("Removed item:", removed_item)
print("After pop():")
print(cart)
````

````python
# 7. Find the index of "Monitor"

monitor_index = cart.index("Monitor")

print("Index of Monitor:", monitor_index)
````

````python
# 8. Count occurrences of "Laptop"

laptop_count = cart.count("Laptop")

print("Number of times Laptop appears:", laptop_count)
````

````python
# 9. Create a copy of the cart

cart_copy = cart.copy()

print("Original cart:")
print(cart)

print("Copied cart:")
print(cart_copy)
````

````python
cart.remove("Monitor")
````

````python
print("Original cart:")
print(cart)

print("Copied cart:")
print(cart_copy)
````

````python
# 10. Reverse the cart

cart.reverse()

print("Cart after reverse():")
print(cart)
````

````python
# 11. Sort the products alphabetically

cart.sort()

print("Cart sorted alphabetically:")
print(cart)
````

## Difference Between Python List Methods

| Method | Purpose | Example |
|---|---|---|
| `append()` | Adds **one item** to the end of a list | `cart.append("Webcam")` |
| `extend()` | Adds **multiple items** from another iterable to the end | `cart.extend(["USB", "Cable"])` |
| `insert()` | Adds an item at a specific index | `cart.insert(2, "USB Hub")` |
| `remove()` | Removes the **first matching value** | `cart.remove("Mouse")` |
| `pop()` | Removes and returns an item, usually the last item | `cart.pop()` |
| `clear()` | Removes **all items** from the list | `cart.clear()` |
| `copy()` | Creates a shallow copy of the list | `cart_copy = cart.copy()` |
| `sort()` | Sorts the list in ascending order | `cart.sort()` |
| `reverse()` | Reverses the order of the list | `cart.reverse()` |

### Important distinction

`append()` adds **one object**:

```python
cart.append("Webcam")
```

`extend()` adds **each item** from another iterable:

```python
cart.extend(["USB Hub", "Cable"])
```

`remove()` works by **value**:

```python
cart.remove("Mouse")
```

`pop()` works by **index** and returns the removed item:

```python
removed = cart.pop()
```

`sort()` changes the original list:

```python
cart.sort()
```

`copy()` creates another list:

```python
cart_copy = cart.copy()
```

`clear()` removes everything:

```python
cart.clear()
```
