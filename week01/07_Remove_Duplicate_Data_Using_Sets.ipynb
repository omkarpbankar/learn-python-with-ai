# Python Sets — Removing Duplicate Data

This exercise demonstrates how a **set** can be used to identify and remove duplicate values from a list.

### Concepts covered
- Lists
- Sets
- Duplicate values
- `set()`
- `list()`
- `len()`
- Unique elements
- Duplicate student names


````python
# Create the original list

numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

# Print the original list
print("Original list:")
print(numbers)
````

````python
# Convert the list into a set

unique_numbers = set(numbers)

print("Set after removing duplicates:")
print(unique_numbers)
````

## Observe the duplicates

The original list contains duplicate values:

- `20` appears twice
- `10` appears twice
- `30` appears twice

A **set only stores unique values**, so duplicate values disappear when the list is converted into a set.

> Note: Sets are unordered collections, so the displayed order of elements should not be relied upon.


````python
# Convert the set back into a list

unique_numbers_list = list(unique_numbers)

print("List after removing duplicates:")
print(unique_numbers_list)
````

````python
# Count the original number of elements

original_count = len(numbers)

print("Number of original elements:", original_count)
````

````python
# Count the number of unique elements

unique_count = len(unique_numbers)

print("Number of unique elements:", unique_count)
````

````python
# Compare original and unique counts

print("Original elements:", len(numbers))
print("Unique elements:", len(unique_numbers))
print("Duplicates removed:", len(numbers) - len(unique_numbers))
````

## Example with Duplicate Student Names

Let's use a list containing duplicate student names.


````python
# Duplicate student names

students = [
    "Rahul",
    "Priya",
    "Aman",
    "Rahul",
    "Neha",
    "Priya",
    "Aman",
    "Rohit"
]

print("Original student list:")
print(students)

# Convert to set
unique_students = set(students)

print("\nUnique student names:")
print(unique_students)

# Convert back to list
unique_students_list = list(unique_students)

print("\nUnique student names as a list:")
print(unique_students_list)

# Count students
print("\nTotal student entries:", len(students))
print("Unique students:", len(unique_students))
````

## Why Are Sets Useful for Duplicate Data?

Sets are useful when you need to quickly identify or remove duplicate values.

### Common use cases

1. **Remove duplicates**
   ```python
   unique_values = set(numbers)
   ```

2. **Count unique values**
   ```python
   len(set(numbers))
   ```

3. **Check whether a value exists**
   ```python
   "Rahul" in unique_students
   ```

4. **Compare collections**

   Sets provide operations such as union, intersection, and difference.

### Real-world examples

Sets are useful in data engineering and data analysis for tasks such as:

- Finding unique customer IDs
- Removing duplicate records
- Finding unique product names
- Identifying unique email addresses
- Comparing two datasets
- Finding common values between datasets

### Important point

A set is **unordered** and contains **only unique elements**. Therefore, converting a list to a set is a convenient way to remove duplicates when the original order is not important.
