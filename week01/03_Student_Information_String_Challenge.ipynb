#### Student Information String Challenge

```python
student = "python programming for data science"
```

```python
# 1. Print the complete string print(student)
print(student)
```

```python
# 2. Print the first character print(student[0])
print(student[0])
```

```python
# 3. Print the last character
print(student[-1])
```

```python
# 4. Print the first 6 characters
print(student[:6])
```

```python
# 5. Print the last 7 characters
print(student[-7:])
```

```python
# 6. Reverse the string using slicing
print(student[::-1])
```

```python
# 7. Convert it to uppercase
print(student.upper())
```

```python
# 8. Convert it to lowercase
print(student.lower())
```

```python
# 9. Convert it to title case
print(student.title())
```

```python
# 10. Count how many times "a" appears
print(student.count("a"))
```

```python
# 11. Find the position of "programming"
print(student.find("programming"))
```

```python
# 12. Replace "data science" with "artificial intelligence"
print(student.replace("data science", "artificial intelligence"))
```

```python
# 13. Split the string into individual words
print(student.split())
```
