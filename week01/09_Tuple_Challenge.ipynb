# Python Tuples — Operations and Immutability

This notebook demonstrates common operations on tuples and explains why tuples are called **immutable**.

### Concepts covered
- Creating a tuple
- `type()`
- Indexing
- Slicing
- `count()`
- `index()`
- `len()`
- Tuple → List
- List → Tuple
- Immutability


````python
# Create the tuple

technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")

print("Tuple:")
print(technologies)
````

````python
# 1. Print the type

print("Type of technologies:")
print(type(technologies))
````

````python
# 2. Print the first item

print("First item:")
print(technologies[0])
````

````python
# 3. Print the last item

print("Last item:")
print(technologies[-1])
````

````python
# 4. Slice the tuple

# Get items from index 1 up to, but not including, index 4
tuple_slice = technologies[1:4]

print("Sliced tuple:")
print(tuple_slice)
````

````python
# 5. Count "Python"

python_count = technologies.count("Python")

print("Number of times Python appears:")
print(python_count)
````

````python
# 6. Find the index of "C++"

cpp_index = technologies.index("C++")

print("Index of C++:")
print(cpp_index)
````

````python
# 7. Find the tuple length

tuple_length = len(technologies)

print("Length of tuple:")
print(tuple_length)
````

````python
# 8. Convert the tuple into a list

technology_list = list(technologies)

print("Tuple converted into list:")
print(technology_list)
print("Type:", type(technology_list))
````

````python
# 9. Add "Go" after converting it into a list

technology_list.append("Go")

print("List after adding Go:")
print(technology_list)
````

````python
# 10. Convert the list back into a tuple

technologies_updated = tuple(technology_list)

print("Final tuple:")
print(technologies_updated)
print("Type:", type(technologies_updated))
````

## Why are tuples called immutable?

A tuple is **immutable**, which means that once a tuple is created, its elements cannot be changed, added, or removed.

For example:

```python
technologies[0] = "Go"
```

This produces a `TypeError` because tuple items cannot be modified.

### If we need to modify a tuple

A common approach is:

1. Convert the tuple into a list.
2. Modify the list.
3. Convert the list back into a tuple.

Example:

```python
technology_list = list(technologies)
technology_list.append("Go")
technologies = tuple(technology_list)
```

The original tuple cannot be modified directly, but the converted list can be changed.

### Key difference

**List:**
- Mutable
- Can be changed after creation
- Uses `[]`

**Tuple:**
- Immutable
- Cannot be changed after creation
- Uses `()`
