## Python Data Type Laboratory

#### PART 1: Creating variables with different data types

```python
# Int Data Type
age = 39 
year = 2026
```

```python
# Float Data Type
salary = 1500000.50 
percentage = 85.75 
```

```python
# String Data Type
name = "Omkar" 
course = "Python Programming" 
```

```python
# Boolean Data Type
likes_python = True 
is_student = False
```

```python
# List Data Type
subjects = ["Python", "SQL", "AWS"]
marks = [85, 90, 88] 
```

```python
# Tuple Data Type
coordinates = (19.07, 72.87) 
months = ("January", "February", "March")
```

```python
# Set Data Type
unique_numbers = {10, 20, 30, 40} 
languages = {"Python", "SQL", "Java"} 
```

```python
# Dictionary Data Type
student = { "name": "Omkar",
            "age": 39, 
            "course": "Python" }

company = { "name": "ABC",
            "city": "Mumbai" }
```

```python
# 2. String to Float 
result2 = float("45.67") 
print("\nfloat('45.67'):", result2) 
print("Data type:", type(result2))
```

```python
# 3. Integer to String 
result3 = str(500) 
print("\nstr(500):", result3) 
print("Data type:", type(result3))
```

```python
# 4. Integer to Boolean 
result4 = bool(0) 
print("\nbool(1):", result4) 
print("Data type:", type(result4))
```

```python
# 5. Tuple to List 
result5 = list((1, 2, 3)) 
print("\nlist((1, 2, 3)):", result5) 
print("Data type:", type(result5))
```

```python
# 6. List to Tuple 
result6 = tuple([1, 2, 3]) 
print("\ntuple([1, 2, 3]):", result6) 
print("Data type:", type(result6))
```

```python
# 7. List to Set 
result7 = set([1, 2, 2, 3]) 
print("\nset([1, 2, 2, 3]):", result7) 
print("Data type:", type(result7))
```
