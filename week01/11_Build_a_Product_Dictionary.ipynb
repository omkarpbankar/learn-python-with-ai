# Python Dictionaries — Product Information

This notebook demonstrates common dictionary operations using a laptop product.

### Concepts covered
- Accessing dictionary values
- Updating values
- Adding new key-value pairs
- Removing dictionary entries
- `keys()`
- `values()`
- `items()`
- Creating a second dictionary


````python
# Create the laptop dictionary

laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}

# Display the product
print("Laptop:")
print(laptop)
````

````python
# Print brand

print("Brand:")
print(laptop["brand"])
````

````python
# Print model

print("Model:")
print(laptop["model"])
````

````python
# Print price

print("Price:")
print(laptop["price"])
````

````python
# Change price

laptop["price"] = 135000

print("Updated price:")
print(laptop["price"])
````

````python
# Add processor information

laptop["processor"] = "Intel Core i7"

print("After adding processor:")
print(laptop)
````

````python
# Add GPU information

laptop["gpu"] = "NVIDIA RTX 4050"

print("After adding GPU:")
print(laptop)
````

````python
# Change RAM to 32GB

laptop["ram"] = "32GB"

print("Updated RAM:")
print(laptop["ram"])
````

````python
# Remove "available"

del laptop["available"]

print("After removing available:")
print(laptop)
````

````python
# Print all keys

print("All keys:")
print(laptop.keys())
````

````python
# Print all values

print("All values:")
print(laptop.values())
````

````python
# Print all items

print("All items:")
print(laptop.items())
````

````python
# Final laptop dictionary

print("Final laptop product:")
print(laptop)
````

## Dictionary Methods Used

| Operation | Purpose | Example |
|---|---|---|
| Access value | Gets the value for a key | `laptop["brand"]` |
| Update value | Changes an existing value | `laptop["price"] = 135000` |
| Add item | Adds a new key-value pair | `laptop["gpu"] = "NVIDIA RTX 4050"` |
| `del` | Removes a key-value pair | `del laptop["available"]` |
| `keys()` | Returns all keys | `laptop.keys()` |
| `values()` | Returns all values | `laptop.values()` |
| `items()` | Returns key-value pairs | `laptop.items()` |

### Important point

Dictionaries store data as **key-value pairs**:

```python
dictionary[key] = value
```

For example:

```python
laptop["ram"] = "32GB"
```

This changes the value associated with the `"ram"` key.
