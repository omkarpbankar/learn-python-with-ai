## Nested List Challenge

```python
students = [ 
    ["Rahul", 21, "Python"], 
    ["Priya", 22, "Data Science"], 
    ["Aman", 20, "Machine Learning"] 
    ]
```

```python
# 1. Print the complete list 
print("Complete List:") 
print(students)
```

```python
# 2. Print Rahul's name 
print("\nRahul's Name:") 
print(students[0][0])
```

```python
# 3. Print Priya's age 
print("\nPriya's Age:") 
print(students[1][1])
```

```python
# 4. Print Aman's course 
print("\nAman's Course:") 
print(students[2][2])
```

```python
# 5. Print the complete record of Priya 
print("\nPriya's Complete Record:") 
print(students[1])
```

```python
# 6. Change Rahul's course to "AI" 
students[0][2] = "AI" 
print("\nAfter Changing Rahul's Course:") 
print(students)
```

```python
# 7. Add another student record manually 
students.append(["Neha", 23, "Python"])
```

```python
# 8. Print the updated nested list 
print("\nUpdated Nested List:") 
print(students)
```
