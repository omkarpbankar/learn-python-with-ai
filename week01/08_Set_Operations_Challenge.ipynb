# Python Sets — Set Operations

This notebook demonstrates set operations using Python and Java student groups.

````python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

print("Python students:", python_students)
print("Java students:", java_students)
````

````python
# Students learning either Python or Java
either_python_or_java = python_students.union(java_students)
print("Either Python or Java:", either_python_or_java)
````

````python
# Students learning both Python and Java
both_python_and_java = python_students.intersection(java_students)
print("Both Python and Java:", both_python_and_java)
````

````python
# Students learning only Python
only_python = python_students.difference(java_students)
print("Only Python:", only_python)
````

````python
# Students learning only Java
only_java = java_students.difference(python_students)
print("Only Java:", only_java)
````

````python
# Students belonging to exactly one group
exactly_one_group = python_students.symmetric_difference(java_students)
print("Exactly one group:", exactly_one_group)
````

````python
# Add a new student
python_students.add("Vikram")
print("After adding Vikram:", python_students)
````

````python
# Remove a student
python_students.remove("Vikram")
print("After removing Vikram:", python_students)
````

````python
# Demonstrate discard()
java_students.discard("Rohit")
print("After discarding Rohit:", java_students)

# No error occurs even though Sachin is not in the set
java_students.discard("Sachin")
print("After discarding missing Sachin:", java_students)
````

## Explanation of Set Methods

| Method | Purpose |
|---|---|
| `union()` | Returns all unique elements from both sets |
| `intersection()` | Returns elements common to both sets |
| `difference()` | Returns elements in the first set but not the second |
| `symmetric_difference()` | Returns elements present in exactly one set |
| `add()` | Adds one element to a set |
| `remove()` | Removes an element; raises `KeyError` if it does not exist |
| `discard()` | Removes an element; does nothing if it does not exist |

### `remove()` vs `discard()`

```python
students.remove("Sachin")
```

Raises `KeyError` if `"Sachin"` is absent.

```python
students.discard("Sachin")
```

Does not raise an error if `"Sachin"` is absent.


````python
# Bonus: equivalent set operators

print("Union:", python_students | java_students)
print("Intersection:", python_students & java_students)
print("Only Python:", python_students - java_students)
print("Exactly one group:", python_students ^ java_students)
````
