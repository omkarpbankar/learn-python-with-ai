# Python Dictionary Practice — Student Record

Practice dictionary access, modification, and common dictionary methods.

````python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}

print("Complete dictionary:")
print(student)
````

````python
print("Student's name:")
print(student["name"])

print("\nStudent's course:")
print(student["course"])
````

````python
print("All keys:")
print(student.keys())

print("\nAll values:")
print(student.values())

print("\nAll key-value pairs:")
print(student.items())
````

````python
# Change marks from 88 to 92
student["marks"] = 92

# Add email
student["email"] = "rahul@example.com"

# Add phone
student["phone"] = "9876543210"

# Remove city
student.pop("city")

print(student)
````

````python
# Use get() to retrieve name
name = student.get("name")
print("Name using get():", name)

# Create a copy
student_copy = student.copy()
print("Copied dictionary:", student_copy)
````

````python
# Demonstrate update()
student.update({
    "marks": 95,
    "course": "Advanced Python"
})

print("After update():")
print(student)
````

## Explanation of Dictionary Methods

| Method | Purpose | Example |
|---|---|---|
| `keys()` | Returns all keys | `student.keys()` |
| `values()` | Returns all values | `student.values()` |
| `items()` | Returns all key-value pairs | `student.items()` |
| `get()` | Retrieves a value safely | `student.get("name")` |
| `update()` | Adds or updates key-value pairs | `student.update({"marks": 95})` |
| `pop()` | Removes a key-value pair and returns its value | `student.pop("city")` |
| `copy()` | Creates a shallow copy of the dictionary | `student.copy()` |

### `get()` vs `[]`

`student["name"]` raises a `KeyError` if the key does not exist.

`student.get("name")` returns `None` by default if the key does not exist, or you can provide a default value:

```python
student.get("email", "Email not available")
```

### `update()`

`update()` can change existing values and add new keys:

```python
student.update({
    "marks": 95,
    "course": "Advanced Python"
})
```

### `pop()`

`pop()` removes the specified key and returns its value:

```python
removed = student.pop("city")
```

### `copy()`

`copy()` creates a separate shallow copy:

```python
student_copy = student.copy()
```

````python
print("Final student dictionary:")
print(student)

print("\nStudent copy:")
print(student_copy)
````
